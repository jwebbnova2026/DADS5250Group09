```python
core_cols = [
    "case_enquiry_id",
    "open_dt",
    "sla_target_dt",
    "closed_dt",
    "on_time",
    "case_status",
    "department",
    "queue",
    "case_title",
    "subject",
    "reason",
    "type",
    "closure_reason"
]

df_simplified = df[core_cols].copy()


# Standardize categorical fields for easier grouping
category_cols = [
    "on_time",
    "case_status",
    "department",
    "queue",
    "case_title",
    "subject",
    "reason",
    "type"
]

for col in category_cols:
    df_simplified[col] = (
        df_simplified[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

print("Dataset shape:", df_simplified.shape)
print(df_simplified.isna().sum().sort_values(ascending=False))
```

```python
date_cols = ["open_dt", "sla_target_dt", "closed_dt"]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], format="mixed").dt.date
```

```python

# Map the department codes
department_mapping = {
    "PWDX": "Public Works",
    "BTDT": "Transportation (parking & traffic)",
    "PARK": "Parks & Recreation",
    "ISD": "Inspectional Services"
}

# Standardize the existing department column
df_simplified["department_code"] = (
    df_simplified["department"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Create the summarized department category
df_simplified["department_category"] = (
    df_simplified["department_code"]
    .map(department_mapping)
    .fillna("Out of scope / review")
)

# Inspect the results

df_simplified[["department", "department_category"]]
```

```python
df_simplified['department_category'].value_counts()
```

```python
keep_departments = [
    "Public Works",
    "Transportation (parking & traffic)",
    "Inspectional Services",
    "Parks & Recreation"
]

df_simplified = df_simplified[df_simplified["department_category"].isin(keep_departments)].copy()
```

```python
date_cols = ["sla_target_dt", "closed_dt"]

for col in date_cols:
    df_simplified[col] = pd.to_datetime(
        df_simplified[col],
        errors="coerce"
    )

# Step 2: Initialize the result as missing
df_simplified["days_late"] = pd.NA

# Step 3: Calculate only when BOTH dates are available
valid_dates = (
    df_simplified["sla_target_dt"].notna()
    & df_simplified["closed_dt"].notna()
)

df_simplified.loc[valid_dates, "days_late"] = (
    df_simplified.loc[valid_dates, "closed_dt"].dt.normalize()
    - df_simplified.loc[valid_dates, "sla_target_dt"].dt.normalize()
).dt.days

# Step 4: Use a nullable integer type
df_simplified["days_late"] = (
    df_simplified["days_late"].astype("Int64")
)
```

```python
df_simplified["open_dt"] = pd.to_datetime(df["open_dt"], format="mixed")
df_simplified["month"] = df_simplified["open_dt"].dt.to_period("M")
```

```python
monthly_demand = (
    df_simplified
    .groupby(["month", "department_category"])
    .size()
    .reset_index(name="request_count")
    .sort_values(["month", "department_category"]))
```

```python
display(monthly_demand)
```

```python
monthly_lateness = (
    df_simplified
    .groupby(["month", "department_category"])
    .agg(
        avg_days_late=("days_late", "mean"),
        late_count=("days_late", lambda x: (x > 0).sum()),
        valid_days_late=("days_late", "count") #number of request where lateness is able to be calculated
    )
    .reset_index()
)

monthly_lateness["late_rate"] = (
    monthly_lateness["late_count"]
    / monthly_lateness["valid_days_late"]
)
```

```python
monthly_summary = monthly_demand.merge(
    monthly_lateness,
    on=["month", "department_category"],
    how="left"
)

monthly_summary.head(12)
```

```python
monthly_summary["month"] = (
    monthly_summary["month"].astype(str).str[:7]
)

def forecast_requests_tool(input_text: str) -> str:
    """Input format: Department | YYYY-MM"""

    department, target_month = [
        part.strip() for part in input_text.split("|", 1)
    ]

    target_calendar_month = target_month[-2:]

    history = monthly_summary[
        (monthly_summary["department_category"] == department)
        & (
            monthly_summary["month"].str[-2:]
            == target_calendar_month
        )
    ]

    if history.empty:
        return f"No historical data found for {department}."

    forecast = math.ceil(history["request_count"].mean())

    return (
        f"Department: {department}\n"
        f"Target month: {target_month}\n"
        f"Forecast requests: {forecast}\n"
        f"Method: Average of historical counts for the "
        f"same calendar month."
    )


forecast_tool = Tool(
    name="MonthlyDemandForecast",
    func=forecast_requests_tool,
    description=(
        "Forecast monthly request volume using historical "
        "counts for the same calendar month. Input format: "
        "Department | YYYY-MM. Example: "
        "Public Works | 2025-01. Use for 2025 forecasts."
    )
)
```

```python
import math

crew_assumptions = {
    "Public Works": {
        "requests_per_day": 12,
        "starting_crews": 50
    },
    "Transportation (parking & traffic)": {
        "requests_per_day": 15,
        "starting_crews": 18
    },
    "Parks & Recreation": {
        "requests_per_day": 6,
        "starting_crews": 10
    },
    "Inspectional Services": {
        "requests_per_day": 4,
        "starting_crews": 16
    }
}

WORKING_DAYS_PER_MONTH = 21


def calculate_crews(department: str, forecast_requests: int) -> str:
    """Calculate monthly crew requirements for a department."""

    assumptions = crew_assumptions[department]

    monthly_capacity_per_crew = (
        assumptions["requests_per_day"]
        * WORKING_DAYS_PER_MONTH
    )

    crews_required = math.ceil(
        forecast_requests / monthly_capacity_per_crew
    )

    return (
        f"Department: {department}\n"
        f"Forecast requests: {forecast_requests}\n"
        f"Capacity per crew per month: {monthly_capacity_per_crew}\n"
        f"Crews required: {crews_required}\n"
        f"Starting crews: {assumptions['starting_crews']}"
    )
```

```python
from langchain_core.tools import Tool

crew_calculator = Tool(
    name="CrewCalculator",
    func=lambda x: calculate_crews(
        x.split("|")[0].strip(),
        int(x.split("|")[1].strip())
    ),
    description="""
    Calculate the number of crews required for a monthly workload.

    Input format:
    Department | Forecast requests

    Example:
    Public Works | 12000

    The tool automatically uses the correct department's
    requests-per-day capacity and assumes 21 working days
    per month.

    Departments supported:
    - Public Works
    - Transportation (parking & traffic)
    - Parks & Recreation
    - Inspectional Services
    """
)
```

```python
from langchain_classic.agents import initialize_agent, AgentType


work_forecaster_instructions = """
You are a workforce planning and budget optimization assistant for municipal service departments.

Your task is to recommend monthly crew allocations based on historical monthly demand, departmental processing capacity, staffing costs, and service-level deadlines.

**Objectives**
1. Maximize the number and percentage of requests completed on time.
2. Keep total annual costs within the specified budget.
3. Account for the financial penalties associated with late requests.
4. Allow some requests to be late when staffing every request on time would exceed the budget or be economically impractical.
5. Recommend cost-effective monthly staffing levels.

**Cost and operational rules**
- Include permanent crew wages, hiring delays, and applicable hiring and release costs.
- Account for monthly crew capacity and departmental demand.
- Apply the specified penalty to requests that miss their service-level deadlines.
- Account for seasonal staffing rules for Parks.
- Track outstanding requests across months where applicable.
- Never assume that late requests disappear at the end of a month.

**Optimization requirements**
- Use the provided Python optimization tool to calculate staffing allocations and costs.
- Do not invent staffing costs, capacity values, demand forecasts, or optimization results.
- Respect the annual budget constraint.
- If the budget cannot support the desired on-time service rate, recommend the best feasible allocation and quantify the expected late requests and penalties.
- Distinguish between forecast estimates and historical observations.

**Output**
Provide:
1. Recommended crew counts by department and month.
2. Projected demand and requests completed on time.
3. Projected late requests and associated penalties.
4. Monthly and annual staffing costs.
5. Total annual cost and remaining budget.
6. Projected on-time service rate.
7. A comparison with the current staffing baseline.
8. An explanation of the trade-offs and any months where demand exceeds available capacity.
"""
tools = [forecast_tool, crew_calculator]

work_forcaster_agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    agent_kwargs={
        "prefix": work_forecaster_instructions
    },
    verbose=True,
    handle_parsing_errors=True )
```

```python
response = work_forcaster_agent.invoke({
    "input": """
    Recommend crew allocations for Public Works,
    Transportation (parking & traffic), Parks & Recreation,
    and Inspectional Services for January 2025.

    Forecast monthly demand for each department, calculate
    required crews, and compare requirements with starting
    staffing levels. Explain which departments may need
    additional crews.
    """
})

pp(response["output"])
```
